> Tip: usa resta y división elemento a elemento. Multiplica por 100 para obtener el porcentaje.


**Para la comprobación**, guarda tu resultado en estas variables (con estos nombres): `month1`, `month2`, `diff`, `pct_cambio`

**Para la comprobación**, guarda tu resultado en estas variables (con estos nombres): `sales_matrix`, `por_producto`, `por_semana`, `idx_mejor`

````python
# Ejercicio 3 — Tu solución
import numpy as np

# 1) Array 2D sales_matrix
sales_matrix = np.array([
    [120, 130, 140, 150],
    [200, 210, 220, 230],
    [90, 100, 110, 120]
    ])

por_producto = np.sum(sales_matrix, axis=1)
por_semana = np.sum(sales_matrix, axis=0)
index_mejor = np.argmax(por_producto)

# 2) Totales por producto, por semana, y producto con mayor venta total

print(por_producto)
print(por_semana)
print(index_mejor)
````
